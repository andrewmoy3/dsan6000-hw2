```python
import boto3, os
url = "s3://dsan6000-wikipedia"

s3 = boto3.client('s3')

for obj in s3.list_objects_v2(Bucket="dsan6000-wikipedia", Prefix="hourly_parquet/")["Contents"]:
    if obj["Key"].endswith(".parquet"):
        s3.download_file("dsan6000-wikipedia", obj["Key"], f"data/{os.path.basename(obj['Key'])}")
```
